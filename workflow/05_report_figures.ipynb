# Report figures

Draws the report figures (Fig. 1, 2, B1, E1–E3) from `report_figures/figure_data.xlsx` and shows them here. No Drive or GPU needed.
Format settings are in `CONFIG` at the top of `report_figures/make_figures.py`.

```
#@title 0. Setup: repository
REPO_URL = "https://github.com/dodokim777-boop/figure.git"  #@param {type:"string"}

import os, subprocess, shutil
REPO_DIR = '/content/' + REPO_URL.rstrip('/').split('/')[-1].removesuffix('.git')
if not os.path.exists(REPO_DIR):
    subprocess.run(['git', 'clone', '-q', REPO_URL, REPO_DIR], check=True)
else:
    subprocess.run(['git', '-C', REPO_DIR, 'pull', '-q'], check=False)
FIG_DIR = os.path.join(REPO_DIR, 'report_figures')
print('ready:', FIG_DIR)
```

```
#@title 1. Draw and show the figures
ONLY = ""  #@param {type:"string"}
#@markdown `ONLY`: leave empty for all figures, or e.g. `E2 E3`.

import glob, sys
from IPython.display import Image, display
cmd = [sys.executable, os.path.join(FIG_DIR, 'make_figures.py')]
if ONLY.strip():
    cmd += ['--only'] + ONLY.split()
subprocess.run(cmd, check=True)
for png in sorted(glob.glob(os.path.join(FIG_DIR, 'output', '*.png'))):
    print(os.path.basename(png))
    display(Image(filename=png, width=900))
```

```
#@title 2. Download all figures (zip)
from google.colab import files
zip_path = shutil.make_archive('/content/report_figures_output', 'zip', os.path.join(FIG_DIR, 'output'))
files.download(zip_path)
```
