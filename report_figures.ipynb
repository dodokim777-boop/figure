# Report figures

Draws the report figures from `figure_data.xlsx` and shows them here. Each figure has its own file (`fig_*.py`); the shared format is in `style.py`.

```
#@title 0. Setup: repository
REPO_URL = "https://github.com/dodokim777-boop/figure.git"  #@param {type:"string"}

import os, subprocess, shutil
REPO_DIR = '/content/' + REPO_URL.rstrip('/').split('/')[-1].removesuffix('.git')
if not os.path.exists(REPO_DIR):
    subprocess.run(['git', 'clone', '-q', REPO_URL, REPO_DIR], check=True)
else:
    subprocess.run(['git', '-C', REPO_DIR, 'pull', '-q'], check=False)
print('ready:', REPO_DIR)
```

```
#@title 1. Draw and show the figures
FIGURES = ""  #@param {type:"string"}
#@markdown Leave empty for all figures, or e.g. `E2 E3` (keys: 1, 2, B1, E1, E2, E3).

import glob, sys
from IPython.display import Image, display
subprocess.run([sys.executable, 'make_all.py'] + FIGURES.split(), cwd=REPO_DIR, check=True)
for png in sorted(glob.glob(os.path.join(REPO_DIR, 'output', '*.png'))):
    print(os.path.basename(png))
    display(Image(filename=png, width=900))
```

```
#@title 2. Download all figures (zip)
from google.colab import files
files.download(shutil.make_archive('/content/report_figures', 'zip', os.path.join(REPO_DIR, 'output')))
```
